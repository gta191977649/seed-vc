# M4Singer WAV File Statistics

This notebook counts how many `.wav` files are available for each artist folder in the dataset and plots the result.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

dataset_path = Path(r"D:\\dev\\research\\dataset\\m4singer")
recursive = True  # Set to False if you only want `.wav` files directly inside each artist folder.

if not dataset_path.exists():
    raise FileNotFoundError(f"Dataset path not found: {dataset_path}")

artist_dirs = sorted(path for path in dataset_path.iterdir() if path.is_dir())
print(f"Dataset path: {dataset_path}")
print(f"Artist folders found: {len(artist_dirs)}")
```

```python
rows = []

for artist_dir in artist_dirs:
    wav_iter = artist_dir.rglob("*.wav") if recursive else artist_dir.glob("*.wav")
    wav_count = sum(1 for _ in wav_iter)
    rows.append({"artist": artist_dir.name, "wav_count": wav_count})

stats_df = pd.DataFrame(rows).sort_values("wav_count", ascending=False).reset_index(drop=True)
stats_df
```

```python
summary = pd.DataFrame(
    {
        "metric": ["artist_folders", "total_wav_files", "max_wav_files", "min_wav_files", "average_wav_files"],
        "value": [
            len(stats_df),
            int(stats_df["wav_count"].sum()),
            int(stats_df["wav_count"].max()) if not stats_df.empty else 0,
            int(stats_df["wav_count"].min()) if not stats_df.empty else 0,
            float(stats_df["wav_count"].mean()) if not stats_df.empty else 0.0,
        ],
    }
)

summary
```

```python
plot_df = stats_df.sort_values("wav_count", ascending=True)

fig_height = max(6, len(plot_df) * 0.25)
plt.figure(figsize=(14, fig_height))
plt.barh(plot_df["artist"], plot_df["wav_count"], color="#4C78A8")
plt.xlabel("Number of WAV files")
plt.ylabel("Artist folder")
plt.title("WAV Files per Artist in M4Singer")
plt.grid(axis="x", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```
